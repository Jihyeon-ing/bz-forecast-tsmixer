```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
symh_pred_bz = np.load("../results/symh_pred_extended_20240303.npy", allow_pickle=True)
symh_pred = np.load("../results/symh_pred_direct_test.npy", allow_pickle=True)
```

```python
symh_pred = symh_pred * (MAX_VALUE - MIN_VALUE) + MIN_VALUE
symh_pred_bz = symh_pred_bz * (MAX_VALUE - MIN_VALUE) + MIN_VALUE
```

```python
total_symh = pd.read_csv("../data/symh_30min_1995_2025.csv", engine="python")
```

```python
plt.figure(figsize=(10, 6))
plt.plot(sym_h[target_start-24:target_start+48], color='dimgray', linewidth=2, marker='D', markersize=4, label='SYM-H (Ground Truth)')
plt.plot([np.nan]*23 + [sym_h[target_start-1]] + list(symh_pred[738]), color='green', linewidth=2, marker='^')
plt.plot([np.nan]*47 + [symh_pred[738][-1]] + list(symh_pred_bz[0]), color='red')
plt.vlines(x=23, ymin=-500, ymax=500, color='black', linestyles='dashed')
plt.vlines(x=47, ymin=-500, ymax=500, color='black', linestyles='dashed')
plt.vlines(x=71, ymin=-500, ymax=500, color='black', linestyles='dashed')
plt.axvspan(23, 47, alpha=0.2, color='forestgreen', label="Direct SYM-H prediction")
plt.axvspan(47, 71, alpha=0.3, color='pink', label="SYM-H prediction from predicted values")
plt.ylim(-250, 50)


plt.ylabel("SYM-H [nT]", fontsize=13, fontweight='bold')
plt.xticks(ticks=[5, 11, 17, 24, 29, 35, 41, 47, 53, 59, 65, 71],
           labels=["-9h", "-6h", "-3h", "+30m", "+3h", "+6h", "+9h", "+12h", "+15h", "+18h", "+21h", "+24h"], fontsize=11)
plt.yticks(fontsize=11)
plt.xlabel("Time", fontsize=13, fontweight='bold')
plt.title(f"Prediction time: {str(pd.to_datetime(time[target_start-1]))[:-3]} UTC", fontsize=13, fontweight='bold')

plt.grid()
plt.legend()
plt.show()
```
